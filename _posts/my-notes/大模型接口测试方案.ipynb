```python
from PIL import Image, ImageDraw, ImageFont

def draw_card(bg_path, text, output_path):
    # 1. 加载你刚生成的底图
    img = Image.open(bg_path)
    draw = ImageDraw.Draw(img)
    
    # 2. 加载中文字体（确保路径下有思源黑体或系统自带字体）
    # macOS 示例: /System/Library/Fonts/PingFang.ttc
    font = ImageFont.truetype("PingFang.ttc", 40) 
    
    # 3. 在右侧留白处压字 (坐标根据你底图的留白位置调整)
    draw.text((550, 150), text, font=font, fill=(255, 255, 255))
    
    img.save(output_path)
    print(f"卡片已合成：{output_path}")

# 调用示例
draw_card("generated_bg.png", "数智能力三问：\n资产化、在线化、复用化", "ajiu_daily_card.png")
```

```python
from PIL import Image, ImageDraw, ImageFont
import textwrap

def draw_ajiu_card(bg_path, title, content, output_path):
    # 1. 加载底图
    img = Image.open(bg_path)
    draw = ImageDraw.Draw(img)
    width, height = img.size

    # 2. 字体设置 (请确保路径正确，Windows建议用 "msyh.ttc", macOS用 "PingFang.ttc")
    try:
        title_font = ImageFont.truetype("PingFang.ttc", 45) # 标题大一点
        content_font = ImageFont.truetype("PingFang.ttc", 30) # 正文适中
    except:
        # 兜底方案：如果找不到字体，加载默认
        title_font = content_font = ImageFont.load_default()

    # 3. 设定右侧留白区域的起始坐标 (根据你之前生成的 40% 留白计算)
    x_offset = int(width * 0.6) + 20 
    y_offset = 120

    # 4. 写入标题
    draw.text((x_offset, y_offset), title, font=title_font, fill=(255, 215, 0)) # 用金色体现质感
    
    # 5. 写入正文 (增加自动换行逻辑)
    y_offset += 80
    lines = textwrap.wrap(content, width=12) # 每行约12个中文字符
    for line in lines:
        draw.text((x_offset, y_offset), line, font=content_font, fill=(255, 255, 255))
        y_offset += 50 # 行间距

    # 6. 加上你的个人 IP 标识
    draw.text((x_offset, height - 80), "—— 阿玖日报 · 把自己产品化", font=content_font, fill=(150, 150, 150))

    img.save(output_path)
    print(f"✅ 卡片已合成：{output_path}")

# 实战调用
draw_ajiu_card(
    "generated_bg.png", 
    "数智能力三问", 
    "资产化、在线化、复用化。通过在线化强制驱动，实现能力的资产化沉淀与跨部门复用。", 
    "ajiu_daily_final.png"
)
```
