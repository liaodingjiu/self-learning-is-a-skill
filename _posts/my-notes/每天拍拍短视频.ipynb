### 我是怎么拍视频的？

有点受不了了，我决定拔刀了。
我今天刷了好几个讲短视频的，乌烟瘴气，看的我脑袋疼，就教了一堆乱七八糟的，这还收费，你怎么好意思？

如果你真的想做短视频，你连听都听不完，你连这个都没耐心，那我觉得你还是别做了，你该干嘛干嘛去。

第一，我们说你的问题，拍短视频是不是不知道拍啥，然后觉得自己有一堆东西说不出来，表达不出来，觉得自己表现力不好，拍完了以后还没有数据也不好，数据好了以后发现没法变现，变现那个事情还不长久。
有没有这个问题？有吧？
我给你做个对比。

第一，**我从来不用提示词**，有人看过我现场演讲，我线下活动也有，我做直播是没有稿子的，是根据现场来决定的，我从来不用提示词。
第二，**我拍短视频之前从不写逐字稿**，从来不写，就拿起来5-10分钟就开始拍，而且质量还都不错，6年了，坚持日更，几乎没断过，到现在为止也没塌房。

这个你可以对比一下，是不是有很多人认为这是能力的问题，我告诉你可能，但是我告诉你，你一个普通人也能做，关键问题是你千万千万别走偏了。

具体怎么做？

就用我一天的流水账，我告诉你我一天怎么干的，你去找我的，方法肯定是干货。

这条视频至少包含十几个干货点，十几个干货点，所以我不知道他到底有多长，而且你看我完全没技巧。

首先第一个就是先从早上开始说，你们说流水账有人听吗？有肯定有。

首先你看我每天早晨送孩子差不多到学校8点，8：15左右，我估计等人家升国旗的时候才进去，我会在外边散散步，这都不是重点。

因为为什么散散步？

因为一个就是我有高血压，散散步这个好点。另外一个就散步的过程中让自己有一些思绪，有些想法。
重点来了，其实这些都是重点，但是这个就是和拍短视频有关系的，就是**我会有一个自己的目标，有一个方向**。

每天我都会想一想今天我干什么，一定是在我那个方向和目标上干什么。
为什么？因为每天的时间特别有限，我每天用来工作的时间就两个多小时，就俩多小时，能干的事很少，所以我要做一件小事，而这个小事是我那个长远方向上的一部分。

好了，散散步回来以后简单收拾一下，10点到12点半，这是我的高效工作时间。这很多老伙计都知道了，一般来说我工作的时候都直播的，30分钟工作，5到10分钟。如果大家有问题，休息的时候给大家聊聊天，大概是这个样子，这个过程中一定是高效工作的。

我的手机是专注模式，谁打电话也打不进来，我该干嘛干嘛。当然这个直播还开着，我也不太关注直播，中间休息的时候跟大家聊。
注意这个问题是关键点了，快结束的那半个小时，最后要结束那个半小时，**我一定要梳理出一张知识卡片出来**。

什么意思？
比如说今天我做了一些事情，这个是今天这个事我干了什么，我要把它梳理成知识卡片出来，一定是找一个实操的，一定有行动的。
如果今天没有就往前梳理，处理以前的，就是**每天最少有一张知识卡片，必须半个小时去梳理它**。

那么这个卡片我是怎么梳理的？注意我的格式，这个，我说过很多遍了，那我再说一遍，我一定是这么梳理的。

**首先第一个，一句话原理**
就是我今天干的这个是什么事。

**第二件事情就是他背后的机制**
他的逻辑推理，为什么做这个事情，以及因果链。

**第三个是什么？**
一个是执行步骤，一个是方法执行步骤。我是怎么做的？
第一步、第二步、第三步、第四步。

**最后一个就是边界和使用条件**
就在什么情况下使用它，什么情况下不用它，注意这个很重要的。

而且这个东西一定不要让AI帮你做，AI可以帮你最后梳理完里边的内容一定是自己的，你能理解吧？
格式它帮你删减一些废话，这都可行，但是内容一定是你自己提供的，千万千万不要让AI帮你，你都没有脑子就我要干一个事，你帮我输出一个东西，我告诉你那个卡片你要不要都没用，别浪费那个时间。

卡片一定是自己的，而且是基于自己的行动的，基于基于自己的理解的这么一张卡片。

好了这是我的关键点，每天这个我每天的工作，这是其中的一个结果。

为什么后面会说？晚上的时候，一般都是晚上，一般把姑娘哄睡了以后，或者他作业做完了，我开始拍一条视频，这个时候我就想，我要抽出哪张卡片出来，可能是今天的，可能是之前的。

那这个卡片，我再多说一句，因为很多人和我不太一样，因为我是有长远的目标的，每天做一点点，你可能没有，那我就教你个最简单的方法。

比如说我看了一本书，我可能就只读了一句话，千万不要泛泛，我就看了一本书，我就只读了一句话，那么这句话我就深入的理解了，我就思考，我怎么样，我怎么把它变成行动，梳理出来，每天一定要有具体的小事去做。

后来我说拍视频，拿出一条视频，拿出一个卡片出来，我说今天和大家聊这个，拿起手机就开始拍，然后，注意为什么不写逐字稿，一会我会解释。

我个人采用的一个比较有效的结构，我有很多结构但是现在我都不用了，我现在基本上就一个结构。
**什么叫结构？**
>- 首先是吸引人的一个钩子
>- 第二个部分是痛点你遇到的问题
>- 第三个是解决方案
>- 第四个是召唤行动，或者一条金句

就这么一个结构。

你看这就是短视频吗？首先一个钩子，你把人吸引进来，你是不是有这个痛点，我告诉你怎么解决。行动大概是这么一个过程，这种视频有一个极大的好处，就是它既能关键的部分是对别人有用，有用就可以保证后面的变现，而且这条视频一定会有用的。

第二个部分是什么？这个内容本身就符合人们喜欢它，通常就会有一定的流量，它不会那么多有流量，但它一定会有流量，是这样一个好短视频。

我为什么说我可以不去不去写逐字稿，因为我的卡片还记得吗？
我的卡片是这样的：
一句话要点，一句话原理。
然后是为什么要做这个事情？逻辑推理因果例机制，机理是不是好？
然后是什么？然后我具体怎么做？
最后是什么？最后是这个边界条件。

但你们注意到了吗？其实它们是一一对应的。
怎么对应？
就是首先钩子，你可以想，我今天要做要讲什么？钩子你可以自己去想，习惯了以后你会发现，你连起都不用起，一张嘴就是，你像我现在都懒得起来，我今天给大家分享个什么东西，直接上来了，我都懒成这样了，当然你最好有一个。

第二个部分是痛点痛点，痛点是不是就是场景，你为什么要做这个事情，你是不是遇到什么问题了，我告诉你，背后的逻辑是什么，是不是就是那个场景，

然后你会发现，解决方案是不是就是前面我这个执行方法，就是我的步骤，我是怎么做的，12345步，对吧？

最后一个，可能是金句，可能是CTA，是我前面的一句话原理就是我前面的，是不是天生就拿过来的？

各位跟大家讲，以前是学建筑的，建筑行业有一个很重要的叫建筑的永恒之道，什么意思？就是挤水，通风的时候应该怎么做？采光的时候应该怎么做？这叫模式，建筑模式。
后来改行做软件，软件有什么？设计模式、架构模式、分析模式，说白了就是在什么时候遇到什么问题该怎么解决，就这么个逻辑。

而这个东西整考是不是和我表达那个东西一样？
所以上来的时候根本就不需要去写稿子，那个东西早很早的时候是我实操的东西，又认真思考过了，还把它变成一张卡片了，那个东西就是我心中的东西。

为什么好多人一定要去背稿子？为什么一定要报稿背稿子？
